# Mostrar árbol de archivos

```python
def mostrar_arbol(
	ruta,
	max_nivel=None,
	prefijo="",
	nivel=0,
	seguir_enlaces=False,
	visitados=None
):
	# Inicializar el conjunto de directorios visitados
	if visitados is None:
		visitados = set()

	# Detener al alcanzar el nivel máximo
	if max_nivel is not None and nivel >= max_nivel:
		return

	# Obtener la ubicación real del directorio
	ruta_real = ruta.resolve()

	# Evitar explorar nuevamente el mismo directorio
	if ruta_real in visitados:
		return

	visitados.add(ruta_real)

	# Obtener y ordenar los elementos
	elementos = sorted(
		ruta.iterdir(),
		key=lambda x: (x.is_file(), x.name.lower())
	)

	for i, elemento in enumerate(elementos):
		ultimo = i == len(elementos) - 1

		conector = "└── " if ultimo else "├── "
		print(prefijo + conector + elemento.name)

		# Identificar enlaces simbólicos y junctions
		es_enlace = elemento.is_symlink()

		if hasattr(elemento, "is_junction"):
			es_enlace = es_enlace or elemento.is_junction()

		# Explorar directorios
		if elemento.is_dir() and (seguir_enlaces or not es_enlace):
			extension = "    " if ultimo else "│   "

			mostrar_arbol(
				elemento,
				max_nivel=max_nivel,
				prefijo=prefijo + extension,
				nivel=nivel + 1,
				seguir_enlaces=seguir_enlaces,
				visitados=visitados
			)
```

## Procesar etiquetas de clase

Existe un archivo llamado list.txt que contiene etiquetas relacionadas con la clasificación de las imágenes.
Podemos obtener su contenido de la siguiente forma:

```python
def cargar_clases(ruta_lista):
	# Cargar las clases desde un archivo de texto
	clases = {}

	with ruta_lista.open("r", encoding="utf-8") as archivo:

		for linea in archivo:

			linea = linea.strip()

			# Ignorar comentarios y líneas vacías
			if not linea or linea.startswith("#"):
				continue

			# Separar los cuatro campos
			image_id, class_id, species_id, breed_id = linea.split()

			# Obtener la raza eliminando el último _número
			raza = image_id.rsplit("_", 1)[0]

			# Guardar los identificadores de la raza
			clases[raza] = {
				"class_id": int(class_id),
				"species_id": int(species_id),
				"breed_id": int(breed_id)
			}

	return clases
```

```python
# Obtener las clases desde el archivo de texto
ruta_lista = DATA / "list.txt"
clases = cargar_clases(ruta_lista)
print(clases)
```

Con este diccionario, podemos obtener las etiquetas de clases de una imagen.

```python
def obtener_clase(image_id, clases):
	# Obtener la información de la clase a partir del identificador de imagen

	raza = image_id.rsplit("_", 1)[0] # Extraer el nombre de la raza

	if raza not in clases:
		raise ValueError(f"Raza no encontrada: {raza}")

	datos = clases[raza]

	# Convertir identificador de especie a nombre
	especies = {
		1: "cat",
		2: "dog"
	}

	return {
		"breed": raza,
		"class_id": datos["class_id"],
		"species_id": datos["species_id"],
		"species": especies[datos["species_id"]],
		"breed_id": datos["breed_id"]
	}
```

```python
# Ejemplo de uso
datos_clase = obtener_clase("Maine_Coon_16", clases)
print(datos_clase)
```

## Procesar atributos de las imágenes

Debemos asegurarnos que las imágenes existan, tengan el mismo tamaño que las máscaras y que los archivos se puedan leer.

```python
def verificar_dimensiones(ruta_imagen, ruta_mascara):
	# Verificar que las dimensiones de la imagen y la máscara sean iguales
	# y que los archivos se puedan leer correctamente

	# Verificar existencia de archivos
	if not ruta_imagen.is_file():
		return 1, None, None # Imagen inexistente

	if not ruta_mascara.is_file():
		return 2, None, None # Máscara inexistente

	try:
		# Obtener dimensiones de la imagen
		with Image.open(ruta_imagen) as imagen:
			ancho, alto = imagen.size

		# Obtener dimensiones de la máscara
		with Image.open(ruta_mascara) as mascara:
			ancho_mascara, alto_mascara = mascara.size
	
	except (OSError, ValueError):
		return 4, None, None # Archivo existente, pero ilegible o inválido

	# Comparar dimensiones
	if (ancho, alto) != (ancho_mascara, alto_mascara):
		return 3, ancho, alto # Dimensiones diferentes

	# Validación exitosa
	return 0, ancho, alto
```

```python
# Ejemplo de uso
image_id = "Maine_Coon_14"
#image_id = "Maine_Coon_15"
raza = image_id.rsplit("_", 1)[0]
ruta_imagen = IMAGES / f"{raza}" / f"{image_id}.jpg"
ruta_mascara = IMAGES / f"{raza}" / f"{image_id}.png"
codigo, ancho, alto = verificar_dimensiones(ruta_imagen, ruta_mascara)

# Imprimir resultados
if codigo == 0:
    print(f"Dimensiones correctas: {ancho}x{alto}")

elif codigo == 1:
    print("Error: no se encontró la imagen")

elif codigo == 2:
    print("Error: no se encontró la máscara")

elif codigo == 3:
    print("Error: las dimensiones no coinciden")

elif codigo == 4:
    print("Error: no se pudo leer alguno de los archivos")

if codigo != 0:
    print(f"Se omite {ruta_imagen.name}: error {codigo}")
```

## Procesar el bounding box

```python
def leer_bounding_box(ruta_xml, ancho, alto):
	# Verifica que el bounding box esté dentro de los límites de la imagen
	# y que el archivo de configuracion xml sea legible.

    # Verificar existencia del XML
    if not ruta_xml.is_file():
        return 1, None # Archivo XML no encontrado

    try:
        raiz = ET.parse(ruta_xml).getroot()

		# Obtener dimensiones declaradas en el XML
        ancho_xml = int(raiz.findtext("size/width"))
        alto_xml = int(raiz.findtext("size/height"))

        # Verificar consistencia con las dimensiones reales
        if (ancho_xml, alto_xml) != (ancho, alto):
            return 5, None # Dimensiones en XML no coinciden con las reales

		# Extraer el bounding box
        bbox = raiz.find("object/bndbox")

        if bbox is None:
            return 2, None # Bounding box no encontrado

        coordenadas = {
            "xmin": int(bbox.findtext("xmin")),
            "ymin": int(bbox.findtext("ymin")),
            "xmax": int(bbox.findtext("xmax")),
            "ymax": int(bbox.findtext("ymax"))
        }

    except (ET.ParseError, OSError, ValueError, TypeError):
        return 4, None # Archivo XML ilegible o inválido

    # Extraer coordenadas
    xmin = coordenadas["xmin"]
    ymin = coordenadas["ymin"]
    xmax = coordenadas["xmax"]
    ymax = coordenadas["ymax"]

    # Validar límites del bounding box
    if not (0 <= xmin < xmax <= ancho):
        return 3, None # Coordenadas del bounding box fuera de los límites de la imagen

    if not (0 <= ymin < ymax <= alto):
        return 3, None # Coordenadas del bounding box fuera de los límites de la imagen

    return 0, coordenadas # Validación exitosa, devolver coordenadas
```

```python
# Ejemplo de uso
image_id = "Sphynx_175"
raza = image_id.rsplit("_", 1)[0]
ruta_imagen = IMAGES / f"{raza}" / f"{image_id}.jpg"
ruta_mascara = IMAGES / f"{raza}" / f"{image_id}.png"
ruta_xml = IMAGES / f"{raza}" / f"{image_id}.xml"

print(f"Procesando imagen: {ruta_imagen.name}")
print(f"Procesando máscara: {ruta_mascara.name}")
print(f"Procesando XML: {ruta_xml.name}")

# Verificar dimensiones de la imagen y la máscara
codigo_dim, ancho, alto = verificar_dimensiones(
    ruta_imagen,
    ruta_mascara
)

if codigo_dim == 0:

    codigo_bbox, bbox = leer_bounding_box(ruta_xml, ancho, alto)

    if codigo_bbox == 0:
        print("Registro válido")
        print("Dimensiones:", ancho, alto)
        print("Bounding box:", bbox)

    elif codigo_bbox == 5:
        print("Error: dimensiones del XML inconsistentes")

    else:
        print(f"Error en XML o bounding box: {codigo_bbox}")
else:
	print(f"Error en validación de imagen y máscara: {codigo_dim}")
```

# Integración del manifest

Debemos procesar la imagen, su máscara y sus metadatos por cada imagen del dataset para conformar las entradas del manifest.

```python
def procesar_imagen(ruta_imagen, directorio_raw, clases):

    image_id = ruta_imagen.stem

    # Construir rutas asociadas
    ruta_mascara = ruta_imagen.with_suffix(".png")
    ruta_xml = ruta_imagen.with_suffix(".xml")

    # Obtener metadatos de la clase
    datos_clase = obtener_clase(image_id, clases)

    # Verificar dimensiones de imagen y máscara
    codigo_dim, ancho, alto = verificar_dimensiones(ruta_imagen, ruta_mascara)

    if codigo_dim != 0:
        return f"DIM_{codigo_dim}", None

    # Leer y validar XML y bounding box
    codigo_bbox, bbox = leer_bounding_box(ruta_xml, ancho, alto)

    if codigo_bbox != 0:
        return f"XML_{codigo_bbox}", None

    # Construir registro del manifest
    registro = {
        "image_id": image_id,
        **datos_clase,
        "image_path": ruta_imagen.relative_to(directorio_raw).as_posix(),
        "mask_path": ruta_mascara.relative_to(directorio_raw).as_posix(),
        "xml_path": ruta_xml.relative_to(directorio_raw).as_posix(),
        "width": ancho,
        "height": alto,
        **bbox
    }

    return 0, registro
```

Se procesa la informacvión que irá en el manifest de manera recursiva.

```python
def generar_manifest(directorio_raw, ruta_lista, ruta_salida):

    # Cargar catálogo de clases una sola vez
    clases = cargar_clases(ruta_lista)

    registros = []
    errores = []

    # Buscar recursivamente archivos JPG
    imagenes = sorted(directorio_raw.rglob("*.jpg"))

    for ruta_imagen in imagenes:

        try:
            codigo, registro = procesar_imagen(ruta_imagen, directorio_raw, clases)

            if codigo == 0:
                registros.append(registro)
                print(f"OK: {ruta_imagen.name}")

            else:
                errores.append({
                    "image_id": ruta_imagen.stem,
                    "error": codigo
                })
                print(f"ERROR: {ruta_imagen.name} → {codigo}")

        except ValueError as error:
            errores.append({
                "image_id": ruta_imagen.stem,
                "error": str(error)
            })
            print(f"ERROR: {ruta_imagen.name} → {error}")

    # Crear DataFrame
    manifest = pd.DataFrame(registros)

    # Guardar CSV
    ruta_salida.parent.mkdir(parents=True, exist_ok=True)
    manifest.to_csv(ruta_salida, index=False)

    print("\nResumen")
    print(f"Imágenes encontradas: {len(imagenes)}")
    print(f"Registros válidos: {len(registros)}")
    print(f"Registros con errores: {len(errores)}")
    print(f"Manifest guardado en: {ruta_salida}")

    return manifest, pd.DataFrame(errores)
```

```python
# Definición de rutas
DATA = PROYECTO / "data" / "raw"
IMAGES = DATA / "images"
ruta_lista = DATA / "list.txt"
ruta_manifest = PROYECTO / "data" / "manifests" / "manifest.csv"

# Integración del manifest
manifest, errores = generar_manifest(IMAGES, ruta_lista, ruta_manifest)

#Visualización de los resultados
print("\nManifest:")
display(manifest)
print("\nErrores:")
display(errores)
```
